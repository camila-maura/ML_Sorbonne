## Kernel Methods

Recall the definition of kernel density estimation:

$$[
f(x_0) = \frac{1}{|X| \sigma^d} \sum_{x \in X} \phi\left(\frac{x_0 - x}{\sigma}\right)
]$$

where (\phi) is a kernel, (\sigma) is a hyperparameter, (X) is the training set, and (d) is the dimension of (X).

### Uniform Kernel

## Nadaraya–Watson Regression

The Nadaraya–Watson estimator allows classification and regression using a method very similar to kernel methods.

Given a training set ({(x_i, y_i)}_{i=1}^n \subset \mathbb{R}^d \times \mathbb{R}), the estimator is defined as:

$$[
f(x) = \frac{\sum_{i=1}^n y_i , \phi\left(\frac{x - x_i}{\sigma}\right)}{\sum_{j=1}^n \phi\left(\frac{x - x_j}{\sigma}\right)}
]$$

Implement the `Nadaraya` class corresponding to this estimator.
